```python
# formatting numbers
# 2 decimal places
salary = 55555445.8967
result = f"{salary:.2f}"
print(result)
```

```python
# formatting numbers
# 4 decimal places
salary = 55555445.8967
result = f"{salary:.4f}"
print(result)
```

```python
# formatting numbers
# Thousand separator
salary = 55555445.89
result = f"{salary:,.2f}"
print(result)
```

```python
# formatting numbers
# Max out the size of the result
salary = 55555445.89
result = f"{salary:14,.2f}"
print(result)
```

```python
# formatting numbers
# Add currency
salary = 55555445.89
result = f"NGN{salary:,.2f}k"
print(result)
```

```python
# Prompt for a number and validate that it is a number until a number is supplied
quantity = input("Enter quantity: ")

while not quantity.isdigit():
    print("Invalid quantity")
    quantity = input("Enter quantity: ")

print(f"Thanks. Valid quantity is {quantity}")
```
