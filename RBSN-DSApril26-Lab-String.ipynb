```python
# Manipulating display of output
# Option-1
firstname, lastname = input('What is your fullname?').split()
age = input('What is your age?')

print('Hello, {} {}. You are {} years old.'.format(firstname, lastname, age))
print("Hello, {} {}. You are {} years old.".format(firstname, lastname, age))
print('Hello, {} {}. You are {} years old.'.format(firstname, age, lastname))
```

```python
# Manipulating display of output
# Option-2
firstname, lastname = input('What is your fullname?').split()
age = input('What is your age?')

print('Hello, {first} {second}. You are {third} years old.'.format(first=firstname, second=lastname, third=age))
print('Hello, {first} {second}. You are {third} years old.'.format(second=lastname, third=age, first=firstname))
```

```python
students =['Ibrahim Hammed', 'Abubakar Balewa']

for item in students:
  # Find the location of " "
  spaceLocation = item.find(" ", 0, len(item))

  # extract the name elements
  firstname = item[:spaceLocation]
  lastname = item[spaceLocation + 1:]

  print(f"Firstname: {firstname}, Lastname: {lastname}")
```
