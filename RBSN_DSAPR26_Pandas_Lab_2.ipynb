```python
# Import needed library
import pandas as pd
```

```python
# Set file location
fileLocation = "https://github.com/IbrahimHammed/RBSN-MSc-Data-Science-2026-Python-Proramming/blob/main/Stock-Sector-Prices-Database.xlsx?raw=true"
```

```python
# Connect to and load data from a file
# Example: .XLSX file
df1 = pd.read_excel(fileLocation)
```

```python
# Get the shape of the dataframe (rows and columns)
df1.shape
```

```python
# Unpivot the dataframe such that Date stands a column next to Ticker which would be headers of other columns and then Value as the last column
df = pd.melt(df1, id_vars=['Date'], var_name='Ticker', value_name='StockPrice')

# Display the first few rows of the melted dataframe to verify
df.head()
```

```python
df.shape
```

```python
# Give a summary of missing values
df.isnull().sum()
```

```python
df_cleaned = df.dropna(subset=['StockPrice'])

# Display the new shape of the dataframe after dropping nulls
print(f"Shape of dataframe after dropping nulls: {df_cleaned.shape}")
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Convert 'Date' column to datetime objects safely
df_cleaned = df_cleaned.copy()
df_cleaned['Date'] = pd.to_datetime(df_cleaned['Date'])
```

Let's pick a specific ticker, for example, 'DANGCEM', to plot its stock price trend over time.

```python
# Select a specific ticker for plotting
selected_ticker = 'DANGCEM'
ticker_df = df_cleaned[df_cleaned['Ticker'] == selected_ticker].sort_values(by='Date')

# Plot the trend of 'StockPrice' over time for the selected ticker
plt.figure(figsize=(14, 7))
sns.lineplot(x='Date', y='StockPrice', data=ticker_df)
plt.title(f'Stock Price Trend for {selected_ticker} Over Time')
plt.xlabel('Date')
plt.ylabel('Stock Price')
plt.grid(True)
plt.tight_layout()
plt.show()
```

This plot shows the 'StockPrice' trend for the selected ticker 'ABBEYBDS' over the entire period available in the dataset. You can change the `selected_ticker` variable in the previous code cell to visualize the trend for any other ticker.
