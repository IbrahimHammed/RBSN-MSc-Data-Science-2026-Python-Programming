```python
# Clean up EPL table
neededHeaders = ['Team', 'Played', 'Won', 'Drawn', 'Lost', 'Goals For', 'Goals Against',
       'Goal Difference', 'Points']

dfEPL = dfEPL[neededHeaders]
dfEPL
```

```python
# Extract the 1st chracter in data series Team into data series Rank
dfEPL['Rank'] = dfEPL['Team'].str[0]
dfEPL
```

```python
# Remove trailing numbers from data series Team
dfEPL['Team'] = dfEPL['Team'].str[1:]
dfEPL
```

```python
# Remove trailing numbers from data series Team starting from row index 9
dfEPL.loc[9:, 'Team'] = dfEPL.loc[9:, 'Team'].str[1:]
dfEPL
```

```python
# Update data series Rank to use values in data series Points in ascending order
dfEPL['Rank'] = dfEPL['Points'].rank(method='min', ascending=False).astype(int)
dfEPL
```
