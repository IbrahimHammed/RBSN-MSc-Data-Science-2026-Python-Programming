```python
# Have fun with students list
# Loop through the list, prompt for Gender, Age, Favourite color and print the result in a tabular form

student_details = []

print("Please provide details for each student:")
for student in students:
    print(f"\nEntering details for student: {student}")
    gender = input(f"Enter Gender for {student}: ")
    age = input(f"Enter Age for {student}: ")
    fav_color = input(f"Enter Favourite Color for {student}: ")

    student_details.append({
        'Student': student,
        'Gender': gender,
        'Age': age,
        'Favourite Color': fav_color
    })

# Prepare data for tabular display
headers = ['Student', 'Gender', 'Age', 'Favourite Color']

# Calculate maximum width for each column
column_widths = {
    header: len(header) for header in headers
}

for detail in student_details:
    for header in headers:
        column_widths[header] = max(column_widths[header], len(str(detail[header])))

# Print header
header_line = " | ".join([h.ljust(column_widths[h]) for h in headers])
print("\nInvoice Table:")
print(header_line)
print("-*-" * (len(header_line) // 3))

# Print rows
for detail in student_details:
    row_line = " | ".join([str(detail[h]).ljust(column_widths[h]) for h in headers])
    print(row_line)
```
