```python
# Example-1: Area of a rectangle
def triangleArea(base:int, height:int)->int:
  area = (base * height) / 2
  return area
```

```python
# Example-3: Return student grade based on score
def getGrade(score:int)->str:
  grade = ''
  if score >= 90:
    grade = "A"
  elif score >= 80:
    grade = "B"
  elif score >= 70:
    grade = "C"
  elif score >= 60:
    grade = "D"
  else:
    grade = "F"
  return grade
```
