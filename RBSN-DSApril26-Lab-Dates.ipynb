```python
# Calculating Age

birthDate = input("When were you born (dd/mm/yyyy)?")
birthDate = dt.datetime.strptime(birthDate, "%d/%m/%Y").date()

today = dt.date.today()

ageInDays = (today - birthDate).days
ageInYears = (today - birthDate).days // 365

print(f"You are {ageInDays:,} days old.")
print(f"You are {ageInYears:,} years old.")
```

```python
# Calculate network days between 2 dates
!pip install python-networkdays

from networkdays import networkdays

startDate = input("Enter start date (dd/mm/yyyy): ")
startDate = dt.datetime.strptime(startDate, "%d/%m/%Y").date()

endDate = input("Enter end date (dd/mm/yyyy): ")
endDate = dt.datetime.strptime(endDate, "%d/%m/%Y").date()

days = networkdays.Networkdays(startDate, endDate)

# Generates a list of valid workdays
workdays = days.networkdays()

print(f"There are {len(workdays)} work days between {startDate} and {endDate}")
```

```python
# Calculate network days between 2 dates
!pip install python-networkdays

from networkdays import networkdays

startDate = dt.date(2026, 1, 1)
endDate = dt.date(2026, 5, 17)

# Set holidays
holidays = {dt.date(2026, 5, 1), dt.date(2026, 1, 1)}

days = networkdays.Networkdays(startDate, endDate, holidays)

# Generates a list of valid workdays
workdays = days.networkdays()

print(f"There are {len(workdays)} work days between {startDate} and {endDate}")
```
