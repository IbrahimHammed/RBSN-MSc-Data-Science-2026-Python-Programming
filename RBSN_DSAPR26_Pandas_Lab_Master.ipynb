```python
# Adding a new column: using a function
def getAgeGroup(age:int) -> str:
  if age < 20:
    return 'Under 20'
  elif age >= 20 and age < 30:
    return '20s'
  elif age >= 30 and age < 40:
    return '30s'
  elif age >= 40 and age < 50:
    return '40s'
  elif age >= 50 and age < 60:
    return '50s'
  elif age >= 60 and age < 70:
    return '60s'
  elif age >= 70 and age < 80:
    return '70s'
  elif age >= 80 and age < 90:
    return '80s'
  elif age >= 90 and age < 100:
    return '90s'
  else:
    return 'Over 100'
```

```python
# The above function is applied to Age column to produce a new column called 'AgeGroup' which is automatically added to the dfRaw
dfRaw['AgeGroup'] = dfRaw['Age'].apply(getAgeGroup)
dfRaw.head()
```

```python
# Aggregating data using GroupBy
# Example 1: Groupby one dimension, one measure
result = dfRaw.groupby('Gender')['CoverageAmount'].sum()
result
```

```python
# Example 2: Groupby one dimension, one measure
result = dfRaw.groupby('MaritalStatus')['CoverageAmount'].sum()
result
```

```python
# Example 3: Groupby multiple dimensions, one measure
result = dfRaw.groupby(['MaritalStatus', 'Gender'])['CoverageAmount'].sum()
result
```

```python
result = dfRaw.groupby(['Gender', 'MaritalStatus'])['CoverageAmount'].sum()
result
```

```python
# Example 4: Groupby multiple dimensions, one measure
dimensions = ['MaritalStatus', 'Gender']
measures = ['CoverageAmount']

result = dfRaw.groupby(dimensions)[measures].sum()
result
```

```python
# Example 5: Groupby multiple dimensions, multiple measures
dimensions = ['MaritalStatus', 'Gender']
measures = ['CoverageAmount', 'PremiumAmount']

result = dfRaw.groupby(dimensions)[measures].sum()
result
```

```python
# Example 6: Groupby multiple dimensions, one measure and many aggregate functions
dimensions = ['MaritalStatus', 'Gender']
measures = ['CoverageAmount']

result = dfRaw.groupby(dimensions)[measures].agg(['sum', 'mean', 'count', 'max', 'min', 'std'])
result
```

```python
# Example 6: Groupby multiple dimensions, one measure and many aggregate functions + row filter(s)
dimensions = ['MaritalStatus', 'Gender']
measures = ['CoverageAmount']

# Apply the age filter before grouping and aggregating
filtered_df = dfRaw[dfRaw['Age'] >= 25]
result = filtered_df.groupby(dimensions)[measures].agg(['sum', 'mean', 'count'])
result
```

### Combining filtering and aggregation into a single line

To perform both the filtering and aggregation in a single line, you can apply the boolean indexing directly to the DataFrame before calling `.groupby()`.

```python
# Example: Groupby multiple dimensions, one measure, many aggregate functions, and row filter(s) in a single line
dimensions = ['MaritalStatus', 'Gender']
measures = ['CoverageAmount']

result_single_line = dfRaw[dfRaw['Age'] >= 25].groupby(dimensions)[measures].agg(['sum', 'mean', 'count'])
result_single_line
```
