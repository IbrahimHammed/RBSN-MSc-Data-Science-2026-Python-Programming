```python
def check_voting_eligibility():
    """
    Prompts the user for their first name and age,
    then determines and prints their voting eligibility.
    """
    print("Welcome to the Voting Eligibility Checker!")

    # 1. Prompt for first name
    first_name = input("Please enter your first name: ")

    # 2. Prompt for age with input validation
    age = -1 # Initialize age to an invalid value to enter the loop
    while True:
        age_input = input(f"Hello {first_name}, please enter your age: ")
        try:
            age = int(age_input)
            if age < 0:
                print("Age cannot be a negative number. Please try again.")
            elif age > 120: # Optional: Add an upper bound for very unrealistic ages
                print("That age seems a bit high. Please enter a realistic age.")
            else:
                break # Exit the loop if age is a valid positive number
        except ValueError:
            print("Invalid input. Please enter a whole number for your age.")

    # 3. Determine voting eligibility (assuming 18 is the voting age)
    print(f"\nThank you, {first_name}!")
    print(f"Your age is: {age}")

    if age >= 18:
        print("Congratulations! You are eligible to vote.")
    else:
        years_to_wait = 18 - age
        print("Unfortunately, you are not yet eligible to vote.")
        if years_to_wait == 1:
            print(f"You will be eligible in {years_to_wait} year.")
        else:
            print(f"You will be eligible in {years_to_wait} years.")
```

```python
check_voting_eligibility()
```
