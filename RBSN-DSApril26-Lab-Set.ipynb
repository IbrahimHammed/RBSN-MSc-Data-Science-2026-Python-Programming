```python
# Create empty sets
nigerianNames = set({})
hausaNames = set({})
igboNames = set({})
yorubaNames = set({})
```

```python
# Add elements to the sets
# Method-1
nigerianNames.add('Olayinka')
nigerianNames.add('Chioma')
nigerianNames.add('Aminu')
nigerianNames
```

```python
# Add elements to the sets
# Method-2
names = ['Yemi', 'Halima', 'Adaora', 'Emeka']

for item in names:
  nigerianNames.add(item)


nigerianNames
```

```python
# Removing elements
# Method-1
nigerianNames.pop()
nigerianNames
```

```python
# Removing elements
# Method-2
nigerianNames.remove('Olayinka')
nigerianNames
```

```python
# Add elements to other sets
x = ['Halima', 'Abubakar', 'Aminu', 'Danladi', 'Zainab']
y = ['Olayinka', 'Yemi', 'Kunle', 'Ajala', 'Labake', 'Yetunde']
z = ['Chinua', 'Ngozi', 'Emeka', 'Adaobi', 'Okonkwo', 'Amarachi']

for item in x:
  hausaNames.add(item)


for item in y:
  yorubaNames.add(item)


for item in z:
  igboNames.add(item)
```

```python
# Union of 2 sets
x = yorubaNames.union(igboNames)
x
```

```python
# Intersection of 2 sets
x = nigerianNames.intersection(hausaNames)
x
```

```python
# Difference of 2 sets
x = nigerianNames.difference(igboNames)
x
```

```python
# Is Superset
x = nigerianNames.issuperset(yorubaNames)
x
```

```python
# Is Subset
x = hausaNames.issubset(nigerianNames)
x
```

```python
# Is disjoint
x = igboNames.isdisjoint(nigerianNames)
x
```
