```python
# Example-3
students = {1:'Ebube', 2:'Ibekwe', 3:'Waziri', 4:'Tolulope', 5: 'Okunu'}
```

```python
# Add more elements
students.update({6: 'Omotola'})
students
```

```python
# Display specific student
students[4]
```

```python
# Displaying all Matric numbers
students.keys()
```

```python
# Displaying all student names
students.values()
```

```python
# Get matric number based on student name
matric_number = None
for key, value in students.items():
    if value == 'Waziri':
        matric_number = key
        break
matric_number
```

```python
# Updating values
students[2] = 'Akpan'
```

```python
# Pop needs the key value
x = students.pop(5)
x
```

```python
# Removing last item
x = students.popitem()
x
```
