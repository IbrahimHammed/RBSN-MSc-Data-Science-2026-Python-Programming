```python
# Import needed library
import pandas as pd
```

```python
# Set file location
fileLocation = "https://github.com/IbrahimHammed/RBSN-MSc-Data-Science-2026-Python-Proramming/blob/main/SalesData.xlsx?raw=true"
```

```python
# Connect to and load data from a file
# Example: .XLSX file
dfRaw = pd.read_excel(fileLocation)
```

```python
# Preview the top 5 rows of the loaded data
dfRaw.head()
```

```python
# Connect to and load clean data from a file
# Example: .XLSX file
ignoreRows = 4
dfClean = pd.read_excel(fileLocation, skiprows= ignoreRows)
dfClean.head()
```

```python
# Check dataframe shape
dfClean.shape
```

```python
# Check RAM consumption
x = dfClean.memory_usage().sum()

# Smarter display in MB
result = x / 1000000
print(f"RAM consumption: {result:.2f} MB")
```
